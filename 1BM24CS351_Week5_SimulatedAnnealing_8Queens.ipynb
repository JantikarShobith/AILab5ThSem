```python
"""
File Name: USN_Week5_SimulatedAnnealing_8_Queens.py
Description: Implementation of Simulated Annealing to solve the 8-Queens Problem.
"""

import math
import random


def calculate_cost(state):
    """Calculates the total number of attacking queen pairs (cost h(S))."""
    cost = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            # Check horizontal conflict (same row)
            if state[i] == state[j]:
                cost += 1
            # Check diagonal conflict (|r_i - r_j| == |i - j|)
            elif abs(state[i] - state[j]) == abs(i - j):
                cost += 1
    return cost


def get_random_neighbor(state):
    """Generates a neighbor state by moving a single randomly chosen queen."""
    neighbor = list(state)
    col = random.randint(0, 7)
    current_row = neighbor[col]

    # Pick a new row different from the current one
    new_row = random.choice([r for r in range(8) if r != current_row])
    neighbor[col] = new_row
    return neighbor


def simulated_annealing_8queens(t_max=100.0, t_min=0.001, alpha=0.995):
    """Solves 8-Queens problem using Simulated Annealing."""
    # Generate initial random state: array of 8 integers between 0 and 7
    current_state = [random.randint(0, 7) for _ in range(8)]
    current_cost = calculate_cost(current_state)

    temperature = t_max
    step = 0

    print("=" * 65)
    print("SIMULATED ANNEALING - 8-QUEENS SEARCH")
    print(f"Initial State : {current_state} | Initial Cost : {current_cost}")
    print("=" * 65)

    while temperature > t_min and current_cost > 0:
        neighbor = get_random_neighbor(current_state)
        neighbor_cost = calculate_cost(neighbor)

        delta_e = neighbor_cost - current_cost

        # Acceptance check via Metropolis criterion
        if delta_e < 0:
            current_state = neighbor
            current_cost = neighbor_cost
        else:
            acceptance_prob = math.exp(-delta_e / temperature)
            if random.random() < acceptance_prob:
                current_state = neighbor
                current_cost = neighbor_cost

        # Cool down
        temperature *= alpha
        step += 1

        if step % 200 == 0:
            print(
                f"Step {step:4d} | Temp: {temperature:7.3f} | Current State: {current_state} | Cost: {current_cost}"
            )

    print("=" * 65)
    if current_cost == 0:
        print(
            f"SUCCESS: Solution found in {step} steps!\nFinal State: {current_state} (Cost: 0)"
        )
    else:
        print(
            f"TERMINATED: Reached minimum temperature.\nBest State Found: {current_state} (Cost: {current_cost})"
        )
    print("=" * 65)

    return current_state, current_cost


def print_board(state):
    """Utility function to display board layout visually."""
    print("\nVisual Board Solution:")
    for row in range(8):
        line = ""
        for col in range(8):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)


if __name__ == "__main__":
    solution, cost = simulated_annealing_8queens()
    if cost == 0:
        print_board(solution)
```
