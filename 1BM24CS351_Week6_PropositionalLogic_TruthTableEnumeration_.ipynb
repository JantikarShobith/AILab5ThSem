```python
"""
File Name: USN_Week6_PropositionalLogic_TruthTableEnumeration.py
Description: Truth-Table Enumeration algorithm for deciding propositional logic entailment.
"""

from itertools import product


# --- Logic Gate Helper Functions ---
def IMPLIES(p, q):
    return (not p) or q


def IFF(p, q):
    return (p and q) or (not p and not q)


def evaluate_kb(model):
    """
    Evaluates the Knowledge Base for a given model assignment.
    Example KB:
      R1: P -> Q
      R2: P
      KB = R1 and R2
    """
    P = model["P"]
    Q = model["Q"]

    rule1 = IMPLIES(P, Q)
    rule2 = P

    return rule1 and rule2


def evaluate_query(model):
    """Evaluates Query (alpha) for a given model assignment."""
    return model["Q"]


def tt_check_all(symbols, model, print_rows=True):
    """
    Recursively builds truth table and evaluates entailment.
    """
    if not symbols:
        kb_val = evaluate_kb(model)
        query_val = evaluate_query(model)

        # Format output row
        p_str = str(model.get("P", False)).ljust(7)
        q_str = str(model.get("Q", False)).ljust(7)
        kb_str = str(kb_val).ljust(9)
        q_val_str = str(query_val).ljust(7)

        # A row passes entailment check if KB is False OR Query is True
        row_entails = (not kb_val) or query_val
        entail_str = str(row_entails)

        if print_rows:
            is_critical = "  <-- CRITICAL ROW" if kb_val else ""
            print(
                f"| {p_str} | {q_str} | {kb_str} | {q_val_str} | {entail_str}{is_critical}"
            )

        if kb_val:
            return query_val
        return True

    P = symbols[0]
    rest = symbols[1:]

    # Branch True then False (Recursive Enumeration)
    model_true = dict(model)
    model_true[P] = True

    model_false = dict(model)
    model_false[P] = False

    result_true = tt_check_all(rest, model_true, print_rows)
    result_false = tt_check_all(rest, model_false, print_rows)

    return result_true and result_false


def tt_entails(symbols):
    """Main function to run TT-Entailment Check."""
    print("=" * 65)
    print("TRUTH TABLE ENUMERATION - PROPOSITIONAL LOGIC")
    print("=" * 65)
    print(f"| {'P':<7} | {'Q':<7} | {'KB':<9} | {'Query':<7} | Entails?")
    print("-" * 65)

    entails = tt_check_all(symbols, {})

    print("=" * 65)
    if entails:
        print("RESULT: KB ENTAILS Query (KB ⊨ α) -> TRUE")
    else:
        print("RESULT: KB DOES NOT ENTAIL Query (KB ⊨ α) -> FALSE")
    print("=" * 65)

    return entails


if __name__ == "__main__":
    # Define symbols involved in KB and Query
    symbols = ["P", "Q"]

    # Run Truth Table Enumeration
    tt_entails(symbols)
```
