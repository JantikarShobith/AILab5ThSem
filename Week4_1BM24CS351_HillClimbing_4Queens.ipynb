```python
def calculate_h(state):
    """
    Calculates the heuristic cost h(S), which is the number of diagonal
    attacking pairs. Since rows are a permutation, horizontal attacks are 0.
    """
    h = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            # Check diagonal attack: |x_i - x_j| == |i - j|
            if abs(state[i] - state[j]) == abs(i - j):
                h += 1
    return h

def get_swap_neighbors(state):
    """
    Generates all 6 possible neighbors by swapping pairs of columns (x_i, x_j) with i < j.
    Returns a list of tuples: (swapped_pair, neighbor_state, neighbor_h)
    """
    neighbors = []
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            # Create a copy and swap row positions of column i and column j
            neighbor = list(state)
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

            h_val = calculate_h(neighbor)
            neighbors.append(((i, j), neighbor, h_val))
    return neighbors

def hill_climbing_4queens_swap(initial_state):
    current_state = list(initial_state)
    iteration = 0

    print("=" * 60)
    print(f"STARTING HILL CLIMBING (ROW-SWAPPING METHOD)")
    print(f"Initial State: {current_state}")
    print("=" * 60)

    while True:
        current_h = calculate_h(current_state)
        print(f"\n--- ITERATION {iteration} ---")
        print(f"Current State S_{iteration}: {current_state} | Cost h(S_{iteration}) = {current_h}")

        # If heuristic is 0, goal reached!
        if current_h == 0:
            print("\n" + "=" * 60)
            print(f"GOAL REACHED! Final Configuration: {current_state} (Cost = 0)")
            print("=" * 60)
            break

        # Generate all 6 swap neighbors
        neighbors = get_swap_neighbors(current_state)
        print("\nEvaluating all 6 swap neighbors:")

        best_neighbor = None
        best_h = current_h
        best_pair = None

        for (i, j), n_state, n_h in neighbors:
            print(f"  - Swap (x_{i}, x_{j}) -> {n_state} | Cost h = {n_h}")

            # Steepest ascent selection with lexicographical tie-breaking for (i, j)
            if n_h < best_h:
                best_h = n_h
                best_neighbor = n_state
                best_pair = (i, j)

        # Check if stuck in a local minimum (no strictly better neighbor)
        if best_neighbor is None or best_h >= current_h:
            print("\nTerminating: Stuck in Local Minimum or Peak (No better neighbor found).")
            break

        print(f"\n>> Selected Best Move: Swap (x_{best_pair[0]}, x_{best_pair[1]}) -> New State: {best_neighbor} with Cost h = {best_h}")
        current_state = best_neighbor
        iteration += 1

# Run the script with the exact initial state from the screenshot
initial_state = [3, 1, 2, 0]
hill_climbing_4queens_swap(initial_state)
```
