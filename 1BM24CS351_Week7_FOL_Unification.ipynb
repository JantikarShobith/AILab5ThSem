```python


def is_variable(expr):
    """Checks if an expression is a variable (starts with lowercase letter)."""
    return isinstance(expr, str) and expr[0].islower()


def occurs_check(var, expr, theta):
    """Prevents unifying a variable with a term containing itself."""
    if var == expr:
        return True
    elif is_variable(expr) and expr in theta:
        return occurs_check(var, theta[expr], theta)
    elif isinstance(expr, list):
        return any(occurs_check(var, arg, theta) for arg in expr)
    return False


def unify_var(var, x, theta):
    """Unifies a variable with an expression."""
    if var in theta:
        return unify(theta[var], x, theta)
    elif is_variable(x) and x in theta:
        return unify(var, theta[x], theta)
    elif occurs_check(var, x, theta):
        return "FAILURE (Occur Check Failed)"
    else:
        new_theta = dict(theta)
        new_theta[var] = x
        return new_theta


def unify(x, y, theta=None):
    """
    Main Unification Function.
    x, y can be string symbols, variables, or nested lists representing compound terms.
    """
    if theta is None:
        theta = {}

    if theta == "FAILURE":
        return "FAILURE"

    # Case 1: Identical expressions
    if x == y:
        return theta

    # Case 2: x is a variable
    elif is_variable(x):
        return unify_var(x, y, theta)

    # Case 3: y is a variable
    elif is_variable(y):
        return unify_var(y, x, theta)

    # Case 4: Both are compound expressions (lists)
    elif isinstance(x, list) and isinstance(y, list):
        if len(x) != len(y):
            return "FAILURE"

        # Unify operator/first elements, then recursively unify remaining arguments
        first_unify = unify(x[0], y[0], theta)
        return unify(x[1:], y[1:], first_unify)

    # Case 5: Cannot unify
    else:
        return "FAILURE"


def test_unification(expr1, expr2):
    """Helper to format test cases."""
    print(f"\nExpr 1 : {expr1}")
    print(f"Expr 2 : {expr2}")
    result = unify(expr1, expr2)
    print(f"Substitution (θ) : {result}")
    print("-" * 55)


if __name__ == "__main__":
    print("=" * 55)
    print("UNIFICATION ALGORITHM FOR FIRST ORDER LOGIC (FOL)")
    print("=" * 55)

    # Test Case 1: Standard unification with variables
    # P(x, Father(x)) AND P(John, y)
    test_unification(["Knows", "John", "x"], ["Knows", "y", "Jane"])

    # Test Case 2: Compound terms
    # Likes(x, f(x)) AND Likes(John, y)
    test_unification(["Likes", "x", ["f", "x"]], ["Likes", "John", "y"])

    # Test Case 3: Failed unification (Mismatched predicates)
    test_unification(["Knows", "John", "x"], ["Sees", "John", "x"])

    # Test Case 4: Occur Check Failure
    test_unification(["P", "x"], ["P", ["f", "x"]])
```
