## 1. Defining the 8-Puzzle State Representation

First, we define how we represent the state of the board, find valid moves (up, down, left, right), and apply those transitions.

```python
import copy

GOAL_STATE = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 0)
)

def find_blank(state):

    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c
    return None

def get_successors(state):

    successors = []
    r, c = find_blank(state)


    moves = [
        (-1, 0, 'Up'),
        (1, 0, 'Down'),
        (0, -1, 'Left'),
        (0, 1, 'Right')
    ]

    for dr, dc, action in moves:
        nr, nc = r + dr, c + dc


        if 0 <= nr < 3 and 0 <= nc < 3:

            new_board = [list(row) for row in state]

            new_board[r][c], new_board[nr][nc] = new_board[nr][nc], new_board[r][c]

            next_state = tuple(tuple(row) for row in new_board)
            successors.append((next_state, action))

    return successors

def depth_limited_search(state, goal, depth, path_states, path_actions):

    if state == goal:
        return path_actions


    if depth <= 0:
        return None


    path_states.add(state)

    for next_state, action in get_successors(state):
        if next_state not in path_states:
            result = depth_limited_search(
                next_state,
                goal,
                depth - 1,
                path_states,
                path_actions + [action]
            )
            if result is not None:
                return result


    path_states.remove(state)
    return None

def iterative_deepening_search(initial_state, goal):

    depth = 0
    while True:
        print(f"Searching at depth limit: {depth}...")
        path_states = set()
        result = depth_limited_search(initial_state, goal, depth, path_states, [])

        if result is not None:
            return result
        depth += 1

if __name__ == "__main__":

    start_board = (
        (1, 2, 3),
        (0, 4, 6),
        (7, 5, 8)
    )

    print("Initial State Layout:")
    for row in start_board:
        print(row)
    print("\nGoal State Layout:")
    for row in GOAL_STATE:
        print(row)
    print("-" * 30)


    solution_moves = iterative_deepening_search(start_board, GOAL_STATE)

    print("-" * 30)
    print(f"Solution Found! Total steps required: {len(solution_moves)}")
    print(f"Sequence of blank space moves: {solution_moves}")
```
